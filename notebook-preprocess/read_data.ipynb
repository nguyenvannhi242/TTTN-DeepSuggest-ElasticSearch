```python
import pandas as pd

df = pd.read_excel("laptop_data.xlsx")

print("Số dòng / số cột:")
print(df.shape)

print("\nDanh sách cột ban đầu:")
for c in df.columns:
    print("-", c)
```

```python
print("\n10 dòng đầu:")
print(df.head(10))
```

```python
print("\nThông tin cấu trúc cột:")
print(df.info())
```

```python
print("\nSố lượng giá trị null mỗi cột:")
print(df.isna().sum())
```

```python
empty_cols = df.columns[df.isna().sum() == len(df)]
print("\nCột hoàn toàn trống:", list(empty_cols))
```

```python
for col in df.columns:
    print(f"\nCột: {col}")
    print(df[col].dropna().astype(str).unique()[:10])
```
